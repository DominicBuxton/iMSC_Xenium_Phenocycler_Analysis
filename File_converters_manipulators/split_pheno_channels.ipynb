```python
sorted_keys = sorted(channel_mapping.keys(), key = lambda k:int(k))

img = tifffile.imread(img_path)

n_channels = len(sorted_keys)
channel_axis = next(i for i, s in enumerate(img.shape) if s == n_channels)

for i, key in enumerate(sorted_keys):
    name = channel_mapping[key]
    channel_img = np.take(img, i, axis = channel_axis)
    out_path = os.path.join(image_folder, f"{name}.tif")
    tifffile.imwrite(out_path, channel_img)
    print(f"saved {out_path}")
```
