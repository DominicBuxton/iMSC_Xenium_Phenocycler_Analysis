# Check if landmarks generated for image registration are in the right paired order

```python
import pandas as pd
import matplotlib.pyplot as plt

# Read the CSV
xen = pd.read_csv(r"D:\Dom\Psoriasis project\4th year data\registered images - first round\xen_pheno_region8\roi2\xen_landmarks.csv")
pheno = pd.read_csv(r"D:\Dom\Psoriasis project\4th year data\registered images - first round\xen_pheno_region8\roi2\pheno_landmarks.csv")
```

```python

# Plot
fig, ax = plt.subplots()

for i, (x, y) in enumerate(zip(pheno['axis-1'], pheno['axis-0'])):
    ax.plot(x, y, 'o', color='steelblue')                        # plot the point
    ax.annotate(str(i), (x, y), textcoords="offset points",      # label with index
                xytext=(5, 5), fontsize=8)

ax.set_xlabel('X (axis-1)')
ax.set_ylabel('Y (axis-0)')
ax.set_title('Landmarks')
plt.tight_layout()
plt.show()
```
